## Import dataset as pandas dataframe

```python
import pandas as pd

df = pd.read_csv("investments_VC.csv", encoding="latin1")

df
```

## Drop useless columns

```python
df.columns = df.columns.str.strip()
df = df.drop(columns=["permalink", "homepage_url"])
df.head()
```

## Drop all rows that contain blank values Except the "state_code" column
The state_code column is excluded since some countries don't contain a "state" system, it is mainly for further classification of data from the US

```python
df = df.dropna(subset=[c for c in df.columns if c != "state_code"])
df
```

# Convert the 'funding_total_usd' column data from string to numerical data and strip whitespaces

```python
df['funding_total_usd'] = (
    df['funding_total_usd']
    .str.strip()                       # Remove surrounding whitespace
    .str.replace(',', '', regex=False) # Remove commas
    .str.replace('-', '0', regex=False)# Replace " - " with 0
    .str.strip()                       # Strip again after replacements
)
df['funding_total_usd'] = pd.to_numeric(df['funding_total_usd'], errors='coerce')
df['funding_total_usd'].unique()
```

# Strip whitespaces from 'market' column name

```python
df['market'] = df['market'].str.strip()
```

# Process the 'category_list' column to contain list of values since it can contain multiple categorical data within it

```python
df['category_list'] = df['category_list'].str.strip('|')
df['category_list'] = df['category_list'].str.split('|')
```

# Process date-time columns to use proper data structure

```python
df['founded_at'] = pd.to_datetime(df['founded_at'], errors='coerce')
df['first_funding_at'] = pd.to_datetime(df['first_funding_at'], errors='coerce')
df['last_funding_at'] = pd.to_datetime(df['last_funding_at'], errors='coerce')
```

# Remove 'founded_at' data that is unrealistic(perhaps corrupted data)

```python
df[df['founded_at'].dt.year < 1900]
```

# Handling missing values to fill categorical NaNs and drop rows of missing numeric values

```python
df['status'].fillna('unknown', inplace=True)
df.dropna(subset=['funding_total_usd'], inplace=True)
```

# Data Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

sns.set_theme(style='whitegrid')
```

```python
df_funded = df[df['funding_total_usd'] > 0]

plt.figure(figsize=(10, 5))
plt.hist(np.log10(df_funded['funding_total_usd']), bins=50, color='steelblue', edgecolor='white')
plt.xlabel('log10(Funding Total USD)')
plt.ylabel('Number of Startups')
plt.title('Distribution of Total Funding (Log Scale)')
plt.tight_layout()
plt.show()
```

```python
top_markets = df['market'].value_counts().head(10)

plt.figure(figsize=(10, 6))
top_markets.sort_values().plot(kind='barh', color='steelblue')
plt.xlabel('Number of Startups')
plt.title('Top 10 Markets by Startup Count')
plt.tight_layout()
plt.show()
```

```python
top_countries = df['country_code'].value_counts().head(10)

plt.figure(figsize=(10, 5))
top_countries.plot(kind='bar', color='coral', edgecolor='white')
plt.xlabel('Country Code')
plt.ylabel('Number of Startups')
plt.title('Top 10 Countries by Startup Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Extract year from founded_at (already parsed as datetime)
df['founded_year'] = df['founded_at'].dt.year

yearly = df.groupby('founded_year')['funding_total_usd'].sum().reset_index()
yearly = yearly[(yearly['founded_year'] >= 1990) & (yearly['founded_year'] <= 2015)]

plt.figure(figsize=(12, 5))
plt.plot(yearly['founded_year'], yearly['funding_total_usd'], marker='o', color='steelblue')
plt.xlabel('Year Founded')
plt.ylabel('Total Funding (USD)')
plt.title('Total VC Funding by Year Founded')
plt.tight_layout()
plt.show()
```

```python
status_counts = df['status'].value_counts()

plt.figure(figsize=(7, 7))
status_counts.plot(kind='pie', autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.ylabel('')
plt.title('Startup Status Distribution')
plt.tight_layout()
plt.show()
```

```python
round_cols = ['seed', 'round_A', 'round_B', 'round_C', 'round_D', 'round_E', 'round_F']
round_totals = df[round_cols].sum()

plt.figure(figsize=(10, 5))
round_totals.plot(kind='bar', color='mediumseagreen', edgecolor='white')
plt.ylabel('Total Funding (USD)')
plt.title('Total Funding Raised per Round Type')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
avg_funding = (
    df.groupby('market')['funding_total_usd']
    .mean()
    .sort_values(ascending=False)
    .head(15)
)

plt.figure(figsize=(10, 7))
avg_funding.sort_values().plot(kind='barh', color='darkorange')
plt.xlabel('Average Funding (USD)')
plt.title('Top 15 Markets by Average Funding')
plt.tight_layout()
plt.show()
```

```python
df_sample = df[df['funding_total_usd'] > 0].sample(2000, random_state=42)

plt.figure(figsize=(9, 6))
plt.scatter(
    df_sample['funding_rounds'],
    np.log10(df_sample['funding_total_usd']),
    alpha=0.3, color='slateblue', edgecolors='none'
)
plt.xlabel('Number of Funding Rounds')
plt.ylabel('log10(Total Funding USD)')
plt.title('Funding Rounds vs. Total Funding')
plt.tight_layout()
plt.show()
```
